### Clustering and Resampling

Let's set up our workspace and use the **Iris dataset** from `scikit-learn`. This dataset is a classic dataset in machine learning and statistics, widely used for clustering tasks. It consists of 150 samples of iris flowers, each belonging to one of three species: Iris setosa, Iris versicolor, and Iris virginica. Here are the key features and characteristics of the dataset:

##### Features:
1. **Sepal Length**: The length of the sepal in centimeters.
2. **Sepal Width**: The width of the sepal in centimeters.
3. **Petal Length**: The length of the petal in centimeters.
4. **Petal Width**: The width of the petal in centimeters.

##### Target Variable:
- **Species**: The species of the iris flower, which can take one of the following values:
  - 0: Iris setosa
  - 1: Iris versicolor

Create plots to visualize the relationships between the features (sepal length, sepal width, petal length, petal width).


```python
import seaborn as sns

# Load the Iris dataset
iris_data = load_iris()

# Convert to DataFrame
iris_df = pd.DataFrame(iris_data.data, columns=iris_data.feature_names)

# Add the target variable (species)
iris_df['species'] = iris_data.target

# Create a pair plot to visualize the relationships between features
sns.pairplot(iris_df, hue='species', diag_kind='kde', markers=["o", "s", "D"])

# Show the plot
plt.show()
```

#### **Question 3:** 
#### Data cleaning

```python
# Initialize the StandardScaler
scaler = StandardScaler()

# Scale the features (excluding the species column)
scaled_features = scaler.fit_transform(iris_df.iloc[:, :-1])

# Create a new DataFrame with scaled features
scaled_iris_df = pd.DataFrame(scaled_features, columns=iris_data.feature_names)

# Add the species column back to the scaled DataFrame
scaled_iris_df['species'] = iris_df['species'].values

# Display the first few rows of the scaled DataFrame
print(scaled_iris_df.head())
```

Why is it important to standardize the features of a dataset before applying clustering algorithms like K-Means? Discuss the implications of using unstandardized data in your analysis.

Given that scales of each parameter differ, the parameter with larger values will influence results to a larger extent than other parameters. This makes the results less accurate and may cause the model to ignore differences of small value parameters. To mitigate that and make every parameter equally weighted, the standartisation allows to bring absolute values to relative ones, with equal scales.

#### **Question 5:** 
#### Bootstrapping 

 Implement bootstrapping on the mean of one of the sepal or petal measurement variables (e.g., Sepal Length, Petal Width) to assess the stability of the mean estimate. Generate 1000 bootstrap samples, calculate the mean for each sample, and compute a 95% confidence interval to evaluate the variability in the estimate.

```python
# Load the Iris dataset from scikit-learn
iris_data = load_iris(as_frame=True).frame

# Print the column names to verify
print(iris_data.columns)

# Define the variable to bootstrap (Sepal Length)
sepal_length = iris_data['sepal length (cm)']

# Set the number of bootstrap samples
n_bootstrap = 1000

# Initialize a list to store the bootstrap means
bootstrap_means = []

# Perform bootstrapping
for _ in range(n_bootstrap):
    # Create a bootstrap sample by sampling with replacement
    bootstrap_sample = np.random.choice(sepal_length, size=len(sepal_length), replace=True)
    
    # Calculate the mean of the bootstrap sample
    bootstrap_means.append(np.mean(bootstrap_sample))

# Convert the list of bootstrap means to a NumPy array
bootstrap_means = np.array(bootstrap_means)

# Calculate the 95% confidence interval (2.5th and 97.5th percentiles)
lower_ci = np.percentile(bootstrap_means, 2.5)
upper_ci = np.percentile(bootstrap_means, 97.5)

# Output the mean, confidence interval
print(f"Bootstrap Mean Estimate: {np.mean(bootstrap_means):.3f}")
print(f"95% Confidence Interval: ({lower_ci:.3f}, {upper_ci:.3f})")

# Plot the distribution of bootstrap means
plt.figure(figsize=(8,6))
sns.histplot(bootstrap_means, kde=True)
plt.axvline(lower_ci, color='red', linestyle='--', label=f'95% CI Lower: {lower_ci:.3f}')
plt.axvline(upper_ci, color='green', linestyle='--', label=f'95% CI Upper: {upper_ci:.3f}')
plt.title('Distribution of Bootstrap Means (Sepal Length)')
plt.xlabel('Mean Sepal Length')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

Reflect on the variability observed in the bootstrapped means and discuss whether the mean of the selected variable appears to be a stable and reliable estimate based on the confidence interval and the spread of the bootstrapped means.

Sepal length sample means are distributed narrowly, with a 95% confidence interval between 5.719 and 5.977. This means that sample means are very close to each other and normally distributed (bell shaped with narrow ends). This in turn means that sampling does not influence the model results significantly, therefore sampling and outliers don't need to be considered.
